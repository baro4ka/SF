```python
# RandomForestClassifier_v1.py
import pandas as pd
import pickle
import os
from sklearn.ensemble import RandomForestClassifier

# 1. Загрузка с заголовком
df = pd.read_csv("data/UCI_Credit_Card.csv")

print("Названия колонок:")
print(df.columns.tolist())
print(f"\nРазмер датасета: {df.shape}")
print("\nПервые 3 строки:")
print(df.head(3))

# 2. Таргет — последняя колонка
target_col = df.columns[-1]
print(f"\nТаргет: '{target_col}'")

X = df.drop(columns=[target_col])
y = df[target_col]

# Убираем ID если он есть
if X.columns[0].lower() in ['id', 'x1']:
    X = X.iloc[:, 1:]
    print("Удалили ID колонку")

# 3. Обучение модели
print("\nОбучаем RandomForestClassifier...")
model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    n_jobs=-1  # используем все ядра
)
model.fit(X, y)

# 4. Сохраняем модель
os.makedirs("models", exist_ok=True)
model_path = "models/rf_default_model.pkl"
with open(model_path, "wb") as f:
    pickle.dump(model, f)

print(f"\nМодель сохранена: {model_path}")
print(f"Количество признаков: {X.shape[1]}")
print(f"Распределение классов:\n{y.value_counts().to_dict()}")

# 5. Проверка качества
from sklearn.metrics import f1_score, recall_score
y_pred = model.predict(X)
print(f"\n Качество на обучении:")
print(f"F1-score: {f1_score(y, y_pred):.4f}")
print(f"Recall: {recall_score(y, y_pred):.4f}")
```
